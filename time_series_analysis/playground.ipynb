```python
import numpy as np
import pandas as pd
import stumpy
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
```

```python
timestamps = pd.date_range(start="2026-07-01", end="2026-07-06", freq="10T")
n_points = len(timestamps)

# Base signal: Diurnal cycle (e.g., tide or daily industrial discharge)
base_signal = np.sin(np.linspace(0, 40 * np.pi, n_points)) 

# Generate 5 sequential sensors with a propagation delay (phase shift)
# Water flows: Sensor 1 -> Sensor 2 -> Sensor 3 -> Sensor 4 -> Sensor 5
data = {
    "Timestamp": timestamps,
    "Sensor_1": base_signal + np.random.normal(0, 0.1, n_points),
    "Sensor_2": np.roll(base_signal, 5) + np.random.normal(0, 0.1, n_points),
    "Sensor_3": np.roll(base_signal, 10) + np.random.normal(0, 0.1, n_points),
    "Sensor_4": np.roll(base_signal, 15) + np.random.normal(0, 0.1, n_points),
    "Sensor_5": np.roll(base_signal, 20) + np.random.normal(0, 0.1, n_points)
}

df = pd.DataFrame(data).set_index("Timestamp")
```
