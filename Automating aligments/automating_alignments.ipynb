```python
from rdflib import Graph,Namespace,URIRef
```

```python
g = Graph()
QUDT = Namespace("http://qudt.org/schema/qudt/")
try:
    g.parse("testdata.ttl", format="turtle")
    print(f"Successfully loaded {len(g)} triples.\n")
except Exception as e:
    print(f"Error parsing the file: {e}")
```

```python
for subject in set(g.subjects()):
    unit_value = g.value(subject, QUDT.unit)
    if unit_value is not None:
       print(f"Subject: {subject}")
       print(f"  └── Found target value: {unit_value}")
       g.parse(URIRef(unit_value))
       conversion_multiplier = g.value(unit_value, QUDT.conversionMultiplier) if g.value(unit_value, QUDT.conversionMultiplier) is not None else 1.0
       conversion_offset = g.value(unit_value, QUDT.conversionOffset) if  g.value(unit_value, QUDT.conversionOffset) is not None else 0.0
       print(f"  └── Found conversion multiplier value: {conversion_multiplier}")
       print(f"  └── Found conversion offset value: {conversion_offset}")
       for predicate, obj in g.predicate_objects(subject):
           print(f"     ├── Predicate: {predicate}")
           print(f"     └── Object:    {obj}")
```
