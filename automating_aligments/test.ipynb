```python
from rdflib import Graph,Namespace,URIRef
```

```python
g = Graph()
deg_c_uri = URIRef("http://qudt.org/vocab/unit/DEG_C")
g.parse(deg_c_uri)
QUDT = Namespace("http://qudt.org/schema/qudt/")
multiplier = g.value(deg_c_uri, QUDT.conversionMultiplier)
offset = g.value(deg_c_uri, QUDT.conversionOffset)
print(f"Unit: {deg_c_uri}")
print("-" * 50)
print(f"Multiplier: {multiplier.toPython() if multiplier else 'Not Found online'}")
print(f"Offset:     {offset.toPython() if offset else 'Not Found online'}")
```
