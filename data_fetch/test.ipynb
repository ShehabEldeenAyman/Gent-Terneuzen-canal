```python
from pywaterinfo import Waterinfo
```

```python
vmm = Waterinfo("vmm")

# 2. Define your parameters
# Group ID 192896 represents 'neerslag 15min'
rainfall_group_id = 192896  
start_date = "2026-05-25"
end_date = "2026-06-01"
```

```python
# 3. Request the data 
# pywaterinfo natively returns data as a parsed Pandas DataFrame
try:
    df_rainfall = vmm.get_timeseries_values(
        timeseriesgroup_id=rainfall_group_id, 
        start=start_date, 
        end=end_date
    )
    
    # 4. Inspect the resulting DataFrame
    print(df_rainfall.head())

except Exception as e:
    print(f"An error occurred: {e}")
```

```python
# Find all individual time series inside the 15-min rainfall group
stations_list = vmm.get_timeseries_list(timeseriesgroup_id=192896)

# Extract individual ts_id for a specific station name or number
# (e.g., searching for the Ertvelde or Koksijde stations)
print(stations_list[["ts_id", "station_name", "station_no"]])
```
